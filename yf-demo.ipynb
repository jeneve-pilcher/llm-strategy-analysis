```python
import yfinance as yf
import pandas as pd

def analyze_financials(company):
    ticker = yf.Ticker(company)

    # Get financial statements
    income = ticker.financials       # Income statement (columns = years)
    balance = ticker.balance_sheet   # Balance sheet
    cashflow = ticker.cashflow       # Cash flow statement

    # Transpose for easier access (rows = years)
    income = income.T
    balance = balance.T

    # Sort years descending (most recent first)
    income.sort_index(ascending=False, inplace=True)
    balance.sort_index(ascending=False, inplace=True)

    try:
        # Revenue Growth (Year-over-Year)
        revenue_growth = (income.loc[income.index[0], 'Total Revenue'] - income.loc[income.index[1], 'Total Revenue']) / income.loc[income.index[1], 'Total Revenue']

        # Profit Margin = Net Income / Total Revenue
        profit_margin = income.loc[income.index[0], 'Net Income'] / income.loc[income.index[0], 'Total Revenue']

        # Gross Margin = Gross Profit / Total Revenue
        gross_margin = income.loc[income.index[0], 'Gross Profit'] / income.loc[income.index[0], 'Total Revenue']

        # ROE = Net Income / Shareholder Equity
        net_income = income.loc[income.index[0], 'Net Income']
        total_equity = balance.loc[balance.index[0], "Total Stockholder Equity"]
        roe = net_income / total_equity

        print(f"📊 Financial Summary for {company}:")
        print(f"Revenue Growth (YoY): {revenue_growth:.2%}")
        print(f"Profit Margin:         {profit_margin:.2%}")
        print(f"Gross Margin:          {gross_margin:.2%}")
        print(f"Return on Equity (ROE):{roe:.2%}")

    except KeyError as e:
        print(f"Missing key in financials: {e}")
    except Exception as e:
        print(f"Error calculating financials: {e}")

# Example: Microsoft
analyze_financials(company)
```
